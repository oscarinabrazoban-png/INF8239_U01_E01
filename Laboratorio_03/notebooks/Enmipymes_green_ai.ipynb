```python
#Cargar y reproducir el dataset
import pandas as pd

df = pd.read_excel(
    "../data/raw/Encuesta-Nacional-a-las-MIPYMES-2023-Base-de-datos.xlsx"
)

print("Dimensiones:", df.shape)

print("\nTipos de datos:")
print(df.dtypes)

print("\nPrimeras filas:")
display(df.head())

print("\nÚltimas filas:")
display(df.tail())

assert not df.empty
```

```python
#Construir la auditoria
audit = pd.DataFrame({
    "tipo": df.dtypes.astype(str),
    "ausentes": df.isna().sum(),
    "porcentaje_ausente": (
        (df.isna().mean() * 100)
        .round(2)
    ),
    "unicos": df.nunique(dropna=False)
}).sort_values(
    "porcentaje_ausente",
    ascending=False
)

print("Duplicados:", df.duplicated().sum())

display(audit)
```

```python
#Limpieza de los espacios en los nombres de las columnas
df.columns = df.columns.str.strip()
```

```python
#Definir target y retirar fugas
TARGET = "FORMALIDAD"

DROP_COLUMNS = []

assert TARGET in df.columns

X = df.drop(columns=[TARGET] + DROP_COLUMNS)
y = df[TARGET]

print(y.value_counts(dropna=False))

assert y.notna().all()
assert y.nunique() >= 2
```

```python
#Verificar las variables categóricas y numéricas
print("Variables numéricas:")
print(X.select_dtypes(include="number").shape[1])

print("Variables categóricas:")
print(X.select_dtypes(exclude="number").shape[1])
```

```python
#Separar tipos y crear el procesamiento de datos
from sklearn.compose import ColumnTransformer
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder, StandardScaler

num_cols = X.select_dtypes(
    include="number"
).columns.tolist()

cat_cols = X.select_dtypes(
    exclude="number"
).columns.tolist()

num_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scale", StandardScaler())
])

cat_pipe = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore"))
])

preprocess = ColumnTransformer([
    ("num", num_pipe, num_cols),
    ("cat", cat_pipe, cat_cols)
])

print(len(num_cols), len(cat_cols))
```

```python
#Dividir, baseline y SVM
from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.svm import SVC
from sklearn.metrics import classification_report, f1_score

Xtr, Xte, ytr, yte = train_test_split(
    X,
    y,
    test_size=.20,
    random_state=42,
    stratify=y
)

dummy = Pipeline([
    ("prep", preprocess),
    ("model", DummyClassifier(strategy="most_frequent"))
])

svm = Pipeline([
    ("prep", preprocess),
    ("model", SVC(
        gamma="scale",
        probability=True,
        random_state=42
    ))
])

for name, model in {
    "dummy": dummy,
    "svm": svm
}.items():

    model.fit(Xtr, ytr)

    pred = model.predict(Xte)

    print(
        name,
        f1_score(
            yte,
            pred,
            average="macro"
        )
    )

print(
    classification_report(
        yte,
        svm.predict(Xte)
    )
)
```

```python
#Inicio de LAB03
#  Catálogo de modelos compatible con salida sparse

from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.pipeline import Pipeline

models = {

    "logistic": Pipeline([
        ("prep", preprocess),
        ("model", LogisticRegression(
            max_iter=2000,
            random_state=42
        ))
    ]),

    "svm_c1": Pipeline([
        ("prep", preprocess),
        ("model", SVC(
            C=1,
            random_state=42
        ))
    ]),

    "svm_linear": Pipeline([
        ("prep", preprocess),
        ("model", SVC(
            kernel="linear",
            random_state=42
        ))
    ]),

    "rf_50": Pipeline([
        ("prep", preprocess),
        ("model", RandomForestClassifier(
            n_estimators=50,
            max_depth=8,
            random_state=42,
            n_jobs=-1
        ))
    ]),

    "rf_100": Pipeline([
        ("prep", preprocess),
        ("model", RandomForestClassifier(
            n_estimators=100,
            max_depth=8,
            random_state=42,
            n_jobs=-1
        ))
    ]),

   "boost": Pipeline([
    ("prep", preprocess),
    ("model", GradientBoostingClassifier(
        random_state=42
    ))
])


}
```

```python
print("Modelos cargados:")
print(list(models.keys()))
```

```python
# Medir tres veces y usar la mediana

from time import perf_counter
from pathlib import Path
import joblib
import numpy as np
import pandas as pd

from sklearn.metrics import (
    f1_score,
    recall_score
)

Path("reports/models").mkdir(
    parents=True,
    exist_ok=True
)

rows = []

for name, model in models.items():

    print(f"\nEntrenando {name}...")

    fit_times = []

    for repetition in range(3):

        start = perf_counter()

        model.fit(Xtr, ytr)

        fit_times.append(
            perf_counter() - start
        )

    start = perf_counter()

    pred = model.predict(Xte)

    predict_ms = (
        perf_counter() - start
    ) * 1000

    model_path = (
        Path("reports/models")
        / f"{name}.joblib"
    )

    joblib.dump(
        model,
        model_path
    )

    rows.append({
        "model": name,
        "f1_macro": f1_score(
            yte,
            pred,
            average="macro"
        ),
        "recall_macro": recall_score(
            yte,
            pred,
            average="macro"
        ),
        "fit_median_s": np.median(
            fit_times
        ),
        "predict_ms": predict_ms,
        "size_kb": (
            model_path.stat().st_size
            / 1024
        )
    })

    print("Completado")

results = pd.DataFrame(rows)

display(
    results.sort_values(
        "f1_macro",
        ascending=False
    )
)
```

```python
#Vamos a verificar is-pareto se genere correctamenrte
def pareto_flags(
    df,
    score="f1_macro",
    cost="fit_time_s"
):
    flags = []

    for _, row in df.iterrows():

        dominated = (
            (
                (df[score] >= row[score])
                &
                (df[cost] <= row[cost])
                &
                (
                    (df[score] > row[score])
                    |
                    (df[cost] < row[cost])
                )
            )
        ).any()

        flags.append(
            not bool(dominated)
        )

    return flags
```

```python
results["is_pareto"] = pareto_flags(
    results,
    cost="fit_median_s"
)

results.sort_values(
    ["is_pareto", "f1_macro"],
    ascending=[False, False]
)
```

```python
# PCA + SVM

from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import f1_score
```

```python
# Trabajaremos solo con variables numéricas

X_num = X[num_cols]

Xtr_num = Xtr[num_cols]
Xte_num = Xte[num_cols]

print(X_num.shape)
```

```python
#Pipeline para PCA y SVM
svm_pca = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scale", StandardScaler()),
    ("pca", PCA(
        n_components=0.95,
        random_state=42
    )),
    ("model", SVC(
        C=1,
        random_state=42
    ))
])
```

```python
#Entrenamiento del modelo PCA + SVM
svm_pca.fit(
    Xtr_num,
    ytr
)

print("Modelo entrenado.")
```

```python
print(
    "Componentes retenidos:",
    svm_pca.named_steps["pca"].n_components_
)
```

```python
#Evaluación del desempeño del modelo PCA + SVM
pred_pca = svm_pca.predict(Xte_num)

f1_pca = f1_score(
    yte,
    pred_pca,
    average="macro"
)

print(
    "F1 Macro PCA:",
    round(f1_pca, 4)
)
```

```python
# t-SNE con dos semillas aleatorias

from sklearn.manifold import TSNE
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Trabajar solo con variables numéricas
X_num = X.select_dtypes(include="number").copy()

# Eliminar columnas completamente vacías
X_num = X_num.dropna(axis=1, how="all")

# Imputar valores faltantes con la mediana
X_num = X_num.fillna(
    X_num.median()
)

# Tomar una muestra de máximo 1000 registros
sample = X_num.sample(
    n=min(1000, len(X_num)),
    random_state=42
)

# Obtener los labels correspondientes
y_sample = y.loc[sample.index]

# Escalar variables
scaled = StandardScaler().fit_transform(sample)

# Verificación de seguridad
print("Shape muestra:", sample.shape)
print("NaN:", np.isnan(scaled).sum())

# Crear dos mapas t-SNE
fig, axes = plt.subplots(
    1,
    2,
    figsize=(12, 5)
)

for ax, seed in zip(axes, [42, 7]):

    emb = TSNE(
        n_components=2,
        perplexity=30,
        random_state=seed,
        init="pca",
        learning_rate="auto"
    ).fit_transform(scaled)

    scatter = ax.scatter(
        emb[:, 0],
        emb[:, 1],
        c=pd.Categorical(y_sample).codes,
        s=12,
        cmap="viridis"
    )

    ax.set_title(
        f"t-SNE seed={seed}"
    )

plt.tight_layout()

plt.savefig(
    "reports/tsne_two_seeds.png",
    dpi=160,
    bbox_inches="tight"
)

plt.show()
```
